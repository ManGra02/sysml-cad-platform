# Publish Box1

Kernel: **SysML** -- check the top right, not Python.

A plain cuboid, and the first model checked against a CAD file this
repository did not generate: `projects/box1/Box1.step`. Stated design:
length 300 mm, width 400 mm, height 500 mm -- one upper bound each.

The model below is `models/box1.sysml` verbatim -- keep the two in step.

````sysml
package Box1 {
    private import ScalarValues::*;
    private import ISQ::*;
    private import SI::*;

    part def Box {
        attribute length : LengthValue;
        attribute width  : LengthValue;
        attribute height : LengthValue;
    }

    requirement def MaxLength {
        doc /* The box must not exceed 300 mm in length. */
        subject box : Box;
        attribute maxLength : LengthValue = 300 [mm];
        require constraint { box.length <= maxLength }
    }

    requirement def MaxWidth {
        doc /* The box must not exceed 400 mm in width. */
        subject box : Box;
        attribute maxWidth : LengthValue = 400 [mm];
        require constraint { box.width <= maxWidth }
    }

    requirement def MaxHeight {
        doc /* The box must not exceed 500 mm in height. */
        subject box : Box;
        attribute maxHeight : LengthValue = 500 [mm];
        require constraint { box.height <= maxHeight }
    }

    part box1 : Box;

    requirement lengthCheck : MaxLength;
    requirement widthCheck  : MaxWidth;
    requirement heightCheck : MaxHeight;

    satisfy lengthCheck by box1;
    satisfy widthCheck  by box1;
    satisfy heightCheck by box1;
}
````

````sysml
%publish Box1
````

If `%publish` returns a project id, the model is on the server. The first
output line must read `http://sysml-api:9000` -- if it says
`sysml2.intercax.com`, it went to the public community server instead.

Then, from the devcontainer:

```bash
python -m sysmlcad.explore -p box1       # dump to out/elements.json
cp out/elements.json projects/box1/elements.json
make verify PROJECT=box1
make verify-api PROJECT=box1
```
