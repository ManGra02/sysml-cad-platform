# Publish Box2

Kernel: **SysML** -- check the top right, not Python.

The second Chili3D export: `projects/box2/Box2.step`. Stated design:
length 100 mm, width 200 mm, height 300 mm -- one upper bound each.

The model below is `models/box2.sysml` verbatim -- keep the two in step.

````sysml
package Box2 {
    private import ScalarValues::*;
    private import ISQ::*;
    private import SI::*;

    part def Box {
        attribute length : LengthValue;
        attribute width  : LengthValue;
        attribute height : LengthValue;
    }

    requirement def MaxLength {
        doc /* The box must not exceed 100 mm in length. */
        subject box : Box;
        attribute maxLength : LengthValue = 100 [mm];
        require constraint { box.length <= maxLength }
    }

    requirement def MaxWidth {
        doc /* The box must not exceed 200 mm in width. */
        subject box : Box;
        attribute maxWidth : LengthValue = 200 [mm];
        require constraint { box.width <= maxWidth }
    }

    requirement def MaxHeight {
        doc /* The box must not exceed 300 mm in height. */
        subject box : Box;
        attribute maxHeight : LengthValue = 300 [mm];
        require constraint { box.height <= maxHeight }
    }

    part box2 : Box;

    requirement lengthCheck : MaxLength;
    requirement widthCheck  : MaxWidth;
    requirement heightCheck : MaxHeight;

    satisfy lengthCheck by box2;
    satisfy widthCheck  by box2;
    satisfy heightCheck by box2;
}
````

````sysml
%publish Box2
````

If `%publish` returns a project id, the model is on the server. The first
output line must read `http://sysml-api:9000` -- if it says
`sysml2.intercax.com`, it went to the public community server instead.

Then, from the devcontainer:

```bash
python -m sysmlcad.explore -p box2       # dump to out/elements.json
cp out/elements.json projects/box2/elements.json
make verify PROJECT=box2
make verify-api PROJECT=box2
```
